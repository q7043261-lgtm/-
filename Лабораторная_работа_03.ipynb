## Задание

Напишите функцию для вычисления суммы квадратов натуральных чисел от **a** до **b**, где **a** и **b** – натуральные числа. Продемоснстрируйте использование объявленной функции.

## Решение

### Объявление функции

```python
def sum_numbers(N):
    s = 0
    for i in range(4, N + 1, 4):
        s += i
    return s
```

### Демонстрация использования функции

```python
N = 10
print(f"Сумма четных чисел, кратных 4, от 1 до {N}: {sum_numbers(N)}")

N = 20
print(f"Сумма четных чисел, кратных 4, от 1 до {N}: {sum_numbers(N)}")

N = 50
print(f"Сумма четных чисел, кратных 4, от 1 до {N}: {sum_numbers(N)}")
```

Решите пример в соответствии с индивидуальным заданием

```python
N = 100
result = sum_numbers(N)

print(f"Сумма четных чисел, кратных 4, от 1 до {N}: {result}")
```

Контрольные вопросы

1. Чем отличается функция с возвратом значения от функции без возврата значения?

Функция с возвратом значения использует оператор return и передает результат своей работы в то место, откуда она была вызвана. Функция без возврата значения выполняет определенные действия, но не возвращает результат.

2. Что такое библиотечная функция?

Библиотечная функция — это готовая функция, которая входит в стандартную библиотеку Python или подключаемую библиотеку. Она предназначена для выполнения определенной задачи и может использоваться в программе без написания собственного алгоритма.

3. Почему библиотечные функции похожи на «черные ящики»?

Библиотечные функции похожи на «черные ящики», потому что пользователю не обязательно знать, как они работают внутри. Достаточно знать название функции, её параметры и результат, который она возвращает.

7. Каким образом функции помогают повторно использовать код в программе?

Функции позволяют один раз написать определенный алгоритм, а затем многократно вызывать его в разных местах программы. Это уменьшает количество повторяющегося кода и упрощает его изменение и исправление.

8. Назовите и опишите две части определения функции.

Определение функции состоит из двух основных частей:

1. Заголовок функции — содержит ключевое слово def, имя функции и её параметры.
2. Тело функции — содержит инструкции, которые выполняются при вызове функции.

Например:

def sum_numbers(N):
    s = 0
    return s

Здесь def sum_numbers(N): — заголовок функции, а инструкции ниже — её тело.

9. Как у функции задать параметры по умолчанию?

Параметр по умолчанию задаётся при объявлении функции с помощью оператора =.

Например:

def greet(name="Пользователь"):
    print("Здравствуйте,", name)

Если при вызове функции значение параметра не указать, будет использовано значение "Пользователь".

10. Что такое рекурсия?

Рекурсия — это способ организации вычислений, при котором функция вызывает сама себя. Для завершения рекурсии должно существовать условие остановки, называемое базовым случаем.

Пример:

def countdown(n):
    if n == 0:
        return
    print(n)
    countdown(n - 1)
